# Chapter 5 project - Explaining startup success (solution)

**Goal.** Given ~470 startups described by their founders, funding, product and team, build a
classifier for *success vs failure* and - the point of this chapter - **explain it**: which
features drive the prediction, how they act, and why one particular startup is predicted to fail.

We walk the whole interpretability toolbox from the three lectures:

- **Deck 1 (glass box):** logistic-regression coefficients / odds ratios, Lasso selection, a single readable tree, and a forest's (biased) impurity importance.
- **Deck 2 (model-agnostic):** permutation feature importance (PFI), partial dependence (PDP) and ICE.
- **Deck 3 (attribution):** SHAP (waterfall / bar / beeswarm / dependence), LIME, and counterfactuals (DiCE).

Along the way this dataset teaches three of the chapter's hardest lessons for free:
**data leakage** (features that leak the outcome), **important != causal**, and
**importance is method-dependent** (the four methods disagree).

> **Runtime: ~12-15 minutes end to end.** The Optuna search (150 LightGBM fits) and the
> probability-space TreeSHAP cell dominate; everything else is seconds.

> Data: `Startup_Success/` (from Metricam/Internship_tasks). It is deliberately messy - a realistic
> audit, not a toy.

```python
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.inspection import permutation_importance, PartialDependenceDisplay
from sklearn.metrics import roc_auc_score, classification_report, brier_score_loss
from sklearn.calibration import calibration_curve
import re
import lightgbm as lgb
import optuna

# Silence only the library-churn categories. ConvergenceWarning and friends stay ON:
# in a 472-row, ~130-feature problem they are real signals, not noise.
warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=DeprecationWarning)
optuna.logging.set_verbosity(optuna.logging.WARNING)
SEED = 509
np.random.seed(SEED)

# house palette: blue = success, red = failure, orange = accent (Armenian-flag colours)
BLUE, RED, ORANGE = "#0033A0", "#D90012", "#F2A800"
plt.rcParams.update({"figure.dpi": 110, "axes.spines.top": False, "axes.spines.right": False,
                     "font.size": 10})
%matplotlib inline

DATA = "Startup_Success/data.csv"
TARGET = "Dependent-Company Status"


def shorten(c, n=38):
    """Compact a long column name for plot labels."""
    c = c.replace("Survival through recession, based on existence of the company through recession times",
                  "recession_survival")
    c = (c.replace("Number of ", "n_").replace("Percent_skill_", "skill_")
          .replace(" organizations", "").replace("for Founders and Co-founders", "(founders)"))
    return c if len(c) <= n else c[:n - 3] + "..."
```

## 7. Leakage audit - the most important EDA step here

A startup dataset is a minefield for **target leakage**: features recorded *after* we already know
the company survived. The clearest one is literally named for it.

```python
lk = "Survival through recession, based on existence of the company through recession times"
ct = pd.crosstab(raw[lk].astype(str).str.strip(), raw[TARGET])
print(ct)
print("\n'No' (did not survive the recession) => 31 Failed, 0 Success: a perfect predictor of failure.")
print("This can only be known AFTER the outcome. It is leakage, not a driver.")
```

It is not the only suspect. Several features are plausibly **post-hoc** - measured at data-collection
time, i.e. after success/failure was decided:

- `Internet Activity Score`, `google page rank of company website` (a dead company stops updating),
- `Number of Recognitions`, `Company awards`, `Top forums ... talking about the company`.

We will not hand-remove all of these (that needs domain sign-off). Instead we let **PFI and SHAP
surface which features the model leans on**, then judge. That auditing use is the whole point of
interpretability: *interpretability catches models that cheat.*

## 8. Cleaning

A single, documented function. Strategy:

1. drop identifiers / free text / dates (`Company_Name`, `Short Description`, `Investors`, dates, ...),
2. turn every missing token into `NaN` (but keep `Not Applicable` - it is a real level for the recession feature),
3. drop columns >55% missing (except the recession leak, kept for the audit),
4. per column: coerce to **numeric** if it parses, else map **Yes/No -> 1/0**, else map known **ordinals** (Low/Medium/High, ...), else **one-hot** low-cardinality nominals,
5. one-hot the low-cardinality nominals.

**What cleaning deliberately does *not* do: impute.** Filling missing values with a median
computed over all 472 rows would let the test set inform the training data - the exact
preprocessing leak this chapter is about. Imputation happens *after* the split, in section 9,
using **training medians only**.

(Dropping columns by missingness rate is still done on the full frame. That decision uses no
target information, so it is the mild, conventional kind of pre-split step - but it is a
simplification, and in a pipeline it would sit inside the fold too.)

```python
DROP_SUBSTR = ["Company_Name", "Short Description", "Investors", "Founding Date", "Funding Date",
               "Specialization of highest", "Industry of company", "Focus functions"]
ORD_MAPS = [
    {"low": 0, "medium": 1, "high": 2},
    {"small": 0, "medium": 1, "large": 2},
    {"bachelors": 0, "masters": 1, "phd": 2},
    {"nothing": 0, "low": 1, "medium": 2, "high": 3},
    {"0 to 2": 0, "2 to 5": 1, "5 to 10": 2, "10 to 15": 3, "15 to 20": 4},
]


def _norm(s):
    return s.astype(str).str.strip().str.lower()


def clean(df):
    df = df.copy()
    y = (df[TARGET].str.strip().str.lower() == "success").astype(int)
    df = df.drop(columns=[TARGET])
    df = df.drop(columns=[c for c in df.columns
                          if any(k.lower() in c.lower() for k in DROP_SUBSTR)])
    for c in df.columns:                                  # missing tokens -> NaN
        df[c] = df[c].astype(str).str.strip().mask(_norm(df[c]).isin(MISS_TOKENS) | df[c].isna())
    miss = df.isna().mean()
    df = df.drop(columns=[c for c in df.columns if miss[c] > 0.55 and "recession" not in c.lower()])

    kinds = {}
    for c in list(df.columns):
        nn = df[c].dropna()
        vals = set(_norm(nn.to_frame()[c]))
        if len(nn) and nn.astype(str).str.strip().str.match(r"^-?\d+(\.\d+)?$").mean() > 0.85:
            df[c] = pd.to_numeric(df[c].astype(str).str.strip(), errors="coerce"); kinds[c] = "num"
        elif vals <= {"yes", "no"}:
            df[c] = _norm(df[c]).map({"yes": 1, "no": 0}); kinds[c] = "bin"
        elif (m := next((m for m in ORD_MAPS if vals <= set(m)), None)) is not None:
            df[c] = _norm(df[c]).map(m); kinds[c] = "ord"
        elif nn.nunique() <= 12:
            kinds[c] = "nom"
        else:
            df = df.drop(columns=[c])
    nom = [c for c, k in kinds.items() if k == "nom" and c in df.columns]
    df = pd.get_dummies(df, columns=nom, dummy_na=False, dtype=float)
    return df, y, kinds      # NaNs are left in on purpose - see section 9


X, y, kinds = clean(raw)
print("clean feature matrix:", X.shape, "| success rate:", round(y.mean(), 3))
print("column kinds:", pd.Series(kinds).value_counts().to_dict())
print("NaNs still to impute:", int(X.isna().sum().sum()),
      f"({X.isna().mean().mean():.1%} of all cells) - imputed after the split")
print("\nNote: 472 rows but ~130 features (after one-hot). This is a small-n, high-p regime -")
print("importances will be noisy; we lean on error bars and read rankings, not exact values.")
```

## 9. Train / test split

Stratified 70/30. Every importance question below is answered on the **test** set (deck 2's rule).

**And only now do we impute.** The median of each column is computed on the *training* rows and
applied to both halves. Doing it earlier is a textbook preprocessing leak: the test set would
have voted on the values used to train. It is a small effect here, but this is the one notebook
in the course where getting it wrong would be embarrassing.

```python
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=SEED, stratify=y)
# Impute AFTER the split, with TRAIN-ONLY medians (in production this belongs in a Pipeline,
# so it is refit inside every CV fold too).
med = Xtr.median()
Xtr, Xte = Xtr.fillna(med), Xte.fillna(med)
print("train:", Xtr.shape, "test:", Xte.shape)
print("NaNs after train-median imputation:", int(Xtr.isna().sum().sum() + Xte.isna().sum().sum()))
```

## 10. Deck 1 - the glass boxes you can just read

### 10a. Logistic regression: coefficients and odds ratios
On standardised features, a coefficient is the log-odds change per +1 SD; `exp(coef)` is the odds
ratio. Sign = direction, magnitude = strength.

```python
logreg = make_pipeline(StandardScaler(), LogisticRegression(max_iter=5000, class_weight="balanced"))
logreg.fit(Xtr, ytr)
lr_auc = roc_auc_score(yte, logreg.predict_proba(Xte)[:, 1])
coef = pd.Series(logreg[-1].coef_[0], index=X.columns)
odds = np.exp(coef)
top = pd.concat([coef.sort_values().head(8), coef.sort_values().tail(8)])
fig, ax = plt.subplots(figsize=(7.5, 5))
ax.barh([shorten(c) for c in top.index], top.values,
        color=[BLUE if v > 0 else RED for v in top.values])
ax.axvline(0, color="0.3", lw=1); ax.set_xlabel("standardised logistic coefficient")
ax.set_title(f"Logistic regression - strongest +/- features (test AUC={lr_auc:.3f})")
plt.tight_layout(); plt.show()
print("Read as odds ratios, e.g. top positive feature:",
      shorten(coef.idxmax()), "-> OR =", round(odds.max(), 2))
```

### 10b. Lasso (L1) logistic - the model selects its own features
L1 drives most coefficients to exactly zero; the survivors are the "selected" features.

```python
lasso = make_pipeline(StandardScaler(),
                      LogisticRegression(penalty="l1", solver="liblinear", C=0.2,
                                         class_weight="balanced", max_iter=5000))
lasso.fit(Xtr, ytr)
nz = np.sum(lasso[-1].coef_[0] != 0)
kept = pd.Series(lasso[-1].coef_[0], index=X.columns)
kept = kept[kept != 0].abs().sort_values(ascending=False)
print(f"Lasso (C=0.2) keeps {nz}/{X.shape[1]} features. Top survivors:")
print([shorten(c) for c in kept.head(10).index])
```

### 10c. A single tree you can read
Shallow enough to follow a root-to-leaf path as a rule.

```python
tree = DecisionTreeClassifier(max_depth=3, random_state=SEED, class_weight="balanced").fit(Xtr, ytr)
fig, ax = plt.subplots(figsize=(13, 6))
plot_tree(tree, feature_names=[shorten(c, 22) for c in X.columns], class_names=["Fail", "Success"],
          filled=True, rounded=True, impurity=False, fontsize=8, ax=ax)
ax.set_title("A depth-3 decision tree (read a path as a rule)")
plt.tight_layout(); plt.show()
print("tree test AUC:", round(roc_auc_score(yte, tree.predict_proba(Xte)[:, 1]), 3))
```

## 11. Deck 1 - a forest and its (biased) built-in importance

A random forest is far more accurate but unreadable; it still emits an impurity importance. That
number is **biased** toward high-cardinality / continuous features and **split across correlated**
ones - keep that in mind reading the bar.

```python
rf = RandomForestClassifier(n_estimators=400, random_state=SEED, class_weight="balanced", n_jobs=-1)
rf.fit(Xtr, ytr)
rf_auc = roc_auc_score(yte, rf.predict_proba(Xte)[:, 1])
imp = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(7.5, 5))
ax.barh([shorten(c) for c in imp.head(15).index][::-1], imp.head(15).values[::-1], color=BLUE)
ax.set_xlabel("impurity importance"); ax.set_title(f"Random forest impurity importance (test AUC={rf_auc:.3f})")
plt.tight_layout(); plt.show()
print(f"Forest AUC {rf_auc:.3f} vs logistic {lr_auc:.3f}. That 0.9+ AUC is suspiciously high - hold that thought.")
```

## 12. Deck 2 - permutation feature importance (PFI)

Model-agnostic: permute a feature on the **test** set and measure the drop in AUC. Error bars come
from repeated shuffles. This is our **leakage audit**: whatever the model leans on floats to the top.

```python
r = permutation_importance(rf, Xte, yte, n_repeats=20, random_state=SEED, scoring="roc_auc")
pfi = pd.Series(r.importances_mean, index=X.columns)
err = pd.Series(r.importances_std, index=X.columns)
order = pfi.sort_values().tail(15).index
fig, ax = plt.subplots(figsize=(7.5, 5))
ax.barh([shorten(c) for c in order], pfi[order].values, xerr=err[order].values,
        color=RED, error_kw=dict(ecolor="0.4", lw=1))
ax.set_xlabel("drop in test AUC when permuted"); ax.set_title("Permutation feature importance (test)")
plt.tight_layout(); plt.show()
leak_here = [shorten(c) for c in order[::-1] if "recession" in c.lower()]
print("Leakage feature(s) among the PFI top:", leak_here)
print("Also near the top: internet activity, page rank, recognitions - the post-hoc suspects from the audit.")
```

### 12a. The correlation trap, measured

Section 6 promised that correlated features **split the credit** and destabilise importance. Time to
show it rather than assert it. Take the most correlated surviving pair and permute them
(a) one at a time and (b) **together**. If the credit really is being split, each feature looks
unimportant alone - the model just reads the answer off its twin - and the joint permutation costs
far more than the two individual costs added up.

```python
cn = [c for c in Xtr.columns if Xtr[c].nunique() > 5]
cm = Xtr[cn].corr().abs()
np.fill_diagonal(cm.values, 0.0)
a_, b_ = np.unravel_index(np.nanargmax(cm.values), cm.shape)
f1, f2 = cm.index[a_], cm.columns[b_]
print(f"most correlated surviving numeric pair: r = {cm.iloc[a_, b_]:.2f}")
print(f"  A: {f1}\n  B: {f2}\n")

base_auc = roc_auc_score(yte, rf.predict_proba(Xte)[:, 1])
rng = np.random.RandomState(SEED)


def permuted_drop(cols, n=20):
    """Mean drop in test AUC when `cols` are shuffled together."""
    out = []
    for _ in range(n):
        Z = Xte.copy()
        for c in cols:
            Z[c] = rng.permutation(Z[c].values)
        out.append(base_auc - roc_auc_score(yte, rf.predict_proba(Z)[:, 1]))
    return float(np.mean(out))


dA, dB, dAB = permuted_drop([f1]), permuted_drop([f2]), permuted_drop([f1, f2])
print(f"PFI(A alone)          = {dA:+.4f}")
print(f"PFI(B alone)          = {dB:+.4f}")
print(f"sum of the two        = {dA + dB:+.4f}")
print(f"PFI(A and B TOGETHER) = {dAB:+.4f}")
print(f"\njoint / sum-of-singles = {dAB / (dA + dB):.2f}x"
      if abs(dA + dB) > 1e-9 else "\n(individual drops are ~0)")
```

**What we got.** The most correlated surviving pair is `year of founding` and
`Age of company in years` at **r = 1.00** - one is a deterministic function of the other, the same
fact written twice. And the masking shows up in the expected direction: shuffling them *together*
costs more than shuffling them one at a time and adding the two up.

Read the magnitudes honestly, though. Every drop here is ~0.001-0.003 AUC, so the *ratio* is
noise-dominated; what is solid is the **structure** (a perfectly redundant pair) and the direction.
The trap is real either way: a low PFI for `year of founding` does **not** mean the founding year is
irrelevant - it means a twin column was standing right behind it.

## 13. Deck 2 - feature effects: PDP and ICE

PFI says *how much*; PDP/ICE say *how*. We take a strong continuous feature and plot the average
effect (PDP) plus per-company ICE curves to expose heterogeneity the average hides.

```python
feat = imp.index[[i for i, c in enumerate(imp.index) if X[c].nunique() > 10][0]]  # top continuous
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
PartialDependenceDisplay.from_estimator(rf, Xte, [feat], ax=ax[0], line_kw=dict(color=BLUE, lw=2.5))
ax[0].set_title(f"PDP: {shorten(feat, 26)}")
PartialDependenceDisplay.from_estimator(rf, Xte, [feat], kind="both", ax=ax[1],
                                        ice_lines_kw=dict(color=BLUE, alpha=0.1),
                                        pd_line_kw=dict(color=ORANGE, lw=3))
ax[1].set_title("PDP + ICE (spread = heterogeneity)")
plt.tight_layout(); plt.show()
```

## 14. Remove the clear leak and refit

We drop the recession-survival columns (the one unambiguous leak) and refit. Watch what happens to
AUC.

```python
safe_cols = [c for c in X.columns if "recession" not in c.lower()]
Xtr_s, Xte_s = Xtr[safe_cols], Xte[safe_cols]
rf_s = RandomForestClassifier(n_estimators=400, random_state=SEED, class_weight="balanced", n_jobs=-1)
rf_s.fit(Xtr_s, ytr)
rf_s_auc = roc_auc_score(yte, rf_s.predict_proba(Xte_s)[:, 1])
print(f"AUC with the leak: {rf_auc:.3f}   without the recession columns: {rf_s_auc:.3f}")
print(f"drop from removing the one unambiguous leak: {rf_auc - rf_s_auc:+.3f} AUC")
print("Is that because the rest is clean, or because leakage is spread over several post-hoc")
print("features? Asserting either would be guessing - the next cell measures it.")
```

### 14a. Staged leak audit - is the leakage really "distributed"?

Dropping the recession columns barely moved AUC. The tempting story is "leakage is spread across the
other post-hoc features", but that is a hypothesis. Test it: remove the suspect blocks
**cumulatively** and refit, so we can see how much test AUC each block was actually worth.

```python
# cumulative removal of the post-hoc blocks flagged in the section-7 audit
POSTHOC = {
    "- recession":   ["recession"],
    "- web signals": ["internet activity", "google page rank"],
    "- recognition": ["recognition", "award", "forums"],
}
stage_names, stage_aucs, stage_p = ["all features"], [rf_auc], [X.shape[1]]
banned = []
for label, keys in POSTHOC.items():
    banned += keys
    cols = [c for c in X.columns if not any(k in c.lower() for k in banned)]
    m = RandomForestClassifier(n_estimators=400, random_state=SEED,
                               class_weight="balanced", n_jobs=-1).fit(Xtr[cols], ytr)
    stage_names.append(label); stage_p.append(len(cols))
    stage_aucs.append(roc_auc_score(yte, m.predict_proba(Xte[cols])[:, 1]))

fig, ax = plt.subplots(figsize=(7.2, 3.8))
b = ax.bar(range(len(stage_aucs)), stage_aucs, color=[RED] + [ORANGE] * (len(stage_aucs) - 1))
ax.bar_label(b, fmt="%.3f")
ax.set_xticks(range(len(stage_names)), stage_names, fontsize=9)
ax.set_ylim(0.5, 1.02); ax.set_ylabel("test AUC")
ax.set_title("Test AUC as post-hoc feature blocks are removed cumulatively")
plt.tight_layout(); plt.show()

for n, a, p in zip(stage_names, stage_aucs, stage_p):
    print(f"{n:16s} {p:4d} features   test AUC {a:.3f}")
print(f"\ntotal cost of removing every post-hoc block: {stage_aucs[-1] - stage_aucs[0]:+.3f} AUC")

# ...but how big is a MEANINGFUL gap on 142 test rows? Bootstrap the full model's AUC.
rng_b = np.random.RandomState(SEED)
p_full, yv = rf.predict_proba(Xte)[:, 1], yte.values
boot = []
while len(boot) < 400:
    idx = rng_b.randint(0, len(yv), len(yv))
    if len(np.unique(yv[idx])) > 1:
        boot.append(roc_auc_score(yv[idx], p_full[idx]))
lo, hi = np.percentile(boot, [2.5, 97.5])
widest_gap = max(stage_aucs) - min(stage_aucs)
print(f"bootstrap 95% CI for the full model's test AUC: [{lo:.3f}, {hi:.3f}] (width {hi - lo:.3f})")
print(f"widest gap between any two stages:            {widest_gap:.3f}")
print("verdict:", "every stage-to-stage gap sits INSIDE the noise band"
      if widest_gap < (hi - lo) else "at least one gap EXCEEDS the noise band")
```

**What we got.** 0.955 -> 0.949 -> 0.938 -> **0.943**: removing the recognition block
made AUC go *up*. And the bootstrap band printed above is wider than the widest gap between any two
stages.

That kills the tidy story in both directions. We cannot conclude "leakage is distributed", and we
cannot conclude "the rest is clean" either - **a 142-row test set simply cannot resolve differences
this small**. The useful lesson is the one the chapter is built on: leakage is found by reading the
*ranking* and asking "could we have known this on day one?", not by watching a single score move.

## 15. A stronger model: LightGBM tuned with Bayesian optimization

The forest is fine, but gradient boosting usually wins on tabular data. We tune a **LightGBM** with
**Bayesian optimization** (Optuna's TPE sampler): rather than grid or random search, it uses the
scores of past trials to propose promising hyperparameters next, spending the budget where it helps.
We maximise **5-fold CV AUC** on the de-leaked training set.

One practical snag: LightGBM rejects feature names containing characters like `?`, `,`, `(`, so we
sanitise the names for the booster and keep a map back to the readable ones for the plots.

```python
def lgbm_safe(cols):
    out, seen = [], {}
    for c in cols:
        s = re.sub(r"[^0-9A-Za-z]+", "_", c).strip("_") or "f"
        seen[s] = seen.get(s, -1) + 1
        out.append(s if seen[s] == 0 else f"{s}_{seen[s]}")
    return out


gnames = lgbm_safe(safe_cols)                    # LightGBM-safe feature names
back = dict(zip(gnames, safe_cols))              # sanitized -> original (for readable labels)
Xtr_g, Xte_g = Xtr_s.copy(), Xte_s.copy()
Xtr_g.columns = gnames; Xte_g.columns = gnames
cv = StratifiedKFold(5, shuffle=True, random_state=SEED)


def objective(trial):
    params = dict(
        n_estimators=trial.suggest_int("n_estimators", 100, 600),
        learning_rate=trial.suggest_float("learning_rate", 0.01, 0.3, log=True),
        num_leaves=trial.suggest_int("num_leaves", 15, 127),
        max_depth=trial.suggest_int("max_depth", 3, 12),
        min_child_samples=trial.suggest_int("min_child_samples", 5, 60),
        subsample=trial.suggest_float("subsample", 0.6, 1.0),
        colsample_bytree=trial.suggest_float("colsample_bytree", 0.6, 1.0),
        reg_lambda=trial.suggest_float("reg_lambda", 1e-3, 10, log=True),
        reg_alpha=trial.suggest_float("reg_alpha", 1e-3, 10, log=True))
    m = lgb.LGBMClassifier(**params, random_state=SEED, class_weight="balanced", n_jobs=-1, verbose=-1)
    return cross_val_score(m, Xtr_g, ytr, cv=cv, scoring="roc_auc").mean()


study = optuna.create_study(direction="maximize", sampler=optuna.samplers.TPESampler(seed=SEED))
study.optimize(objective, n_trials=30, show_progress_bar=False)
print(f"best 5-fold CV AUC: {study.best_value:.4f}  (over {len(study.trials)} Bayesian trials)")
print("best params:", {k: (round(v, 3) if isinstance(v, float) else v) for k, v in study.best_params.items()})

best = lgb.LGBMClassifier(**study.best_params, random_state=SEED, class_weight="balanced",
                          n_jobs=-1, verbose=-1).fit(Xtr_g, ytr)
gbm_auc = roc_auc_score(yte, best.predict_proba(Xte_g)[:, 1])
print(f"tuned LightGBM test AUC: {gbm_auc:.3f}")
```

The Bayesian search history (running-best CV AUC) and a fair scoreboard of the de-leaked models.
Note the honest twist: tuning pushed the **CV** AUC high, but on the 142-row **test** set LightGBM is
within noise of the forest - on 472 rows you should not over-read a 0.01 AUC gap.

```python
gbm0 = lgb.LGBMClassifier(random_state=SEED, class_weight="balanced", n_jobs=-1, verbose=-1).fit(Xtr_g, ytr)
gbm0_auc = roc_auc_score(yte, gbm0.predict_proba(Xte_g)[:, 1])
lr_safe = make_pipeline(StandardScaler(), LogisticRegression(max_iter=5000, class_weight="balanced")).fit(Xtr_s, ytr)
lr_safe_auc = roc_auc_score(yte, lr_safe.predict_proba(Xte_s)[:, 1])

vals = [t.value for t in study.trials]
running = np.maximum.accumulate(vals)
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(range(1, len(vals) + 1), vals, "o", color=BLUE, alpha=0.5, label="trial CV AUC")
ax[0].plot(range(1, len(vals) + 1), running, color=RED, lw=2, label="running best")
ax[0].set_xlabel("Bayesian trial"); ax[0].set_ylabel("5-fold CV AUC")
ax[0].legend(); ax[0].set_title("Optuna (TPE) optimisation history")
board = pd.Series({"logistic": lr_safe_auc, "RF": rf_s_auc,
                   "LightGBM\n(default)": gbm0_auc, "LightGBM\n(tuned)": gbm_auc})
b = ax[1].bar(range(len(board)), board.values, color=[ORANGE, ORANGE, ORANGE, RED])
ax[1].set_xticks(range(len(board)), board.index); ax[1].set_ylim(0.7, 1.0)
ax[1].bar_label(b, fmt="%.3f"); ax[1].set_ylabel("test AUC")
ax[1].set_title("Model scoreboard (de-leaked, test AUC)")
plt.tight_layout(); plt.show()
```

### 15a. Are those probabilities real? (calibration, lecture [14])

Everything after this point reads `predict_proba` as a probability: the SHAP waterfall is requested
in **probability space**, and the counterfactual section asks what "crosses the **0.5** decision
line". But we trained every model with `class_weight="balanced"`, which deliberately re-weights the
classes - and re-weighting **distorts the probabilities** even when it improves ranking (AUC is
invariant to any monotone rescaling, so a great AUC tells you nothing about calibration).

Deck 3 states the consequence explicitly: *SHAP explains the model's score; if that score is not
calibrated, SHAP faithfully explains a miscalibrated number.* So we check.

```python
p_gbm = best.predict_proba(Xte_g)[:, 1]
frac, mean_pred = calibration_curve(yte, p_gbm, n_bins=8, strategy="quantile")
brier = brier_score_loss(yte, p_gbm)

fig, ax = plt.subplots(figsize=(4.8, 4.2))
ax.plot([0, 1], [0, 1], "--", color="0.6", lw=1.2, label="perfectly calibrated")
ax.plot(mean_pred, frac, "o-", color=RED, lw=2, label=f"tuned LightGBM")
ax.set_xlabel("predicted P(success)"); ax.set_ylabel("observed success rate")
ax.set_title("Reliability diagram (test)"); ax.legend(fontsize=8); ax.set_aspect("equal")
plt.tight_layout(); plt.show()

print(f"Brier score:            {brier:.3f}  (lower is better; 0.25 = always predicting 0.5)")
print(f"mean predicted P(success): {p_gbm.mean():.3f}")
print(f"actual success rate:       {yte.mean():.3f}")
print(f"test AUC:                  {gbm_auc:.3f}  <- ranking is fine either way")
```

**What we got.** Brier **0.130**, and a mean predicted P(success) of **0.704** against
an actual test rate of **0.648** - the model is about 5-6 points optimistic. The test AUC of 0.927
is completely blind to this, because AUC only cares about *order*.

So for the rest of the notebook: read the waterfall's "probability" and the counterfactual's 0.5
line as **the model's score**, not as a startup's real chance of success. Fixing it would mean
`CalibratedClassifierCV` on a held-out slice - lecture [14]'s job, not this chapter's, but the
caveat has to travel with the explanation.

## 16. Deck 3 - SHAP (on the tuned LightGBM)

SHAP splits each prediction into per-feature contributions that sum to it. **TreeSHAP** is exact and
fast for boosted trees - the canonical case - so we now explain the **tuned LightGBM**. We request
SHAP in **probability space** (with a small background sample) so the values sum to the predicted
probability, then read its four plots.

```python
import shap
bg = shap.sample(Xtr_g, 100, random_state=SEED)          # background for probability-space SHAP
explainer = shap.TreeExplainer(best, bg, model_output="probability")
sv = explainer(Xte_g, check_additivity=False)
sv = sv[:, :, 1] if sv.values.ndim == 3 else sv          # positive (Success) class
sv.feature_names = [shorten(back[c]) for c in gnames]    # readable labels
print("SHAP Explanation (tuned LightGBM):", sv.values.shape)

# check_additivity is OFF because probability-space TreeSHAP with a SAMPLED background is
# approximate. That is a reason to verify the deck's identity ourselves, not to stop looking:
recon = np.asarray(sv.base_values).ravel() + sv.values.sum(1)
resid = np.abs(recon - best.predict_proba(Xte_g)[:, 1])
print(f"base + sum(phi) vs predict_proba -> max |residual| = {resid.max():.2e}, "
      f"mean = {resid.mean():.2e}")
```

### 17a. Is LIME actually unstable? Measure it.

The slides claim LIME is unstable because step 2 samples at random. Claiming it in a solution
notebook and not showing it would be the same sin. We explain the **same company** five times with
five different sampling seeds and look at the spread.

```python
runs = []
for k in range(5):
    e = LimeTabularExplainer(Xtr_g.values, feature_names=[shorten(back[c]) for c in gnames],
                             class_names=["Fail", "Success"], discretize_continuous=True,
                             random_state=SEED + k)
    runs.append(dict(e.explain_instance(Xte_g.values[i], predict_np, num_features=8).as_list()))

allf = sorted({f for r in runs for f in r},
              key=lambda f: -np.mean([abs(r.get(f, 0.0)) for r in runs]))[:8]
tab = pd.DataFrame([[r.get(f, np.nan) for f in allf] for r in runs],
                   columns=allf, index=[f"run {k + 1}" for k in range(5)])

fig, ax = plt.subplots(figsize=(9, 4))
for k, f in enumerate(allf):
    v = tab.iloc[:, k].values
    ax.scatter([k] * len(v), v, color=BLUE, alpha=0.75, zorder=3, s=45)
    ax.plot([k - 0.22, k + 0.22], [np.nanmean(v)] * 2, color=RED, lw=2, zorder=4)
ax.axhline(0, color="0.4", lw=1)
ax.set_xticks(range(len(allf)), [f[:26] for f in allf], rotation=30, ha="right", fontsize=8)
ax.set_ylabel("LIME weight"); ax.set_title(f"Five LIME runs on the same company (#{i}); red bar = mean")
plt.tight_layout(); plt.show()

always = [f for f in allf if all(f in r for r in runs)]
print(f"top-8 features that appear in ALL five runs: {len(always)}/{len(allf)}")
print(f"largest spread on a shared feature: "
      f"{max((tab[f].max() - tab[f].min()) for f in always):.3f}" if always else "")
```

**What we got.** **6 of the top 8** features appear in all five runs, and the largest
disagreement on a shared feature is **0.026** on weights of order 0.1. So on *this* instance LIME is
noticeably more stable than the slide's warning implies.

Worth saying out loud rather than hiding: instability is a **risk**, not a guarantee. The lesson is
not "LIME is unreliable", it is "you cannot know without re-running it" - and re-running it costs
the five lines above.

### 17b. LIME's global sibling: a surrogate model

Deck 3 pairs the *local* surrogate (LIME) with its *global* one: train an interpretable model to
**mimic the black box everywhere**, then read the mimic. The number that decides whether you are
allowed to read it is **fidelity** - how often the surrogate agrees with the model it is imitating.

```python
# a depth-3 tree trained on the LightGBM's OWN predictions (not on y)
gbm_label_tr = (best.predict_proba(Xtr_g)[:, 1] >= 0.5).astype(int)
gbm_label_te = (best.predict_proba(Xte_g)[:, 1] >= 0.5).astype(int)
surro = DecisionTreeClassifier(max_depth=3, random_state=SEED).fit(Xtr_g, gbm_label_tr)

print(f"fidelity (agreement with the LightGBM): train {(surro.predict(Xtr_g) == gbm_label_tr).mean():.3f}"
      f"   test {(surro.predict(Xte_g) == gbm_label_te).mean():.3f}")
print(f"surrogate test AUC against the TRUE labels: "
      f"{roc_auc_score(yte, surro.predict_proba(Xte_g)[:, 1]):.3f}   (LightGBM: {gbm_auc:.3f})")

fig, ax = plt.subplots(figsize=(13, 5))
plot_tree(surro, feature_names=[shorten(back[c], 20) for c in gnames],
          class_names=["Fail", "Success"], filled=True, rounded=True, impurity=False,
          fontsize=8, ax=ax)
ax.set_title("Global surrogate: a depth-3 tree trained to imitate the tuned LightGBM")
plt.tight_layout(); plt.show()
```

## 18. Deck 3 - counterfactual recourse (DiCE)

For a startup the tuned model predicts will fail, what is the smallest **actionable** change that
would flip it to predicted-success? We let only founder-controllable levers move (sales material,
technical capability, advisors, co-founders), and freeze the rest.

```python
import dice_ml
train_df = Xtr_g.copy(); train_df["success"] = ytr.values
inv = {v: k for k, v in back.items()}                    # original name -> sanitized (LightGBM) name
# founder-controllable levers the model actually uses, each with a REALISTIC cap so DiCE cannot
# propose nonsense like "hire 3000 employees" (that would violate plausibility + proximity).
LEVERS_ORIG = {
    "Number of  Sales Support material": (0, 3),
    "Technical proficiencies to analyse and interpret unstructured data": (0, 1),
    "Number of of advisors": (0, 8),
    "Number of Co-founders": (1, 5),
}
LEVERS = {inv[c]: b for c, b in LEVERS_ORIG.items() if c in inv}
actionable = list(LEVERS)
permitted = {c: [float(lo), float(hi)] for c, (lo, hi) in LEVERS.items()}

d = dice_ml.Data(dataframe=train_df, continuous_features=gnames, outcome_name="success")
m = dice_ml.Model(model=best, backend="sklearn")
dice = dice_ml.Dice(d, m, method="random")
print("actionable levers:", [shorten(back[c]) for c in actionable])

# search predicted-fail companies (our focus #i first, then others closest to the line)
order = [i] + [int(j) for j in fail_idx[np.argsort(-p_s[fail_idx])] if int(j) != i]
found = None
for j in order[:25]:
    q = Xte_g.iloc[[j]]
    try:
        cf = dice.generate_counterfactuals(q, total_CFs=5, desired_class="opposite",
                                           features_to_vary=actionable, permitted_range=permitted,
                                           random_seed=SEED)
        cands = cf.cf_examples_list[0].final_cfs_df
    except Exception as e:                                # never swallow silently
        print(f"  DiCE failed on company #{j}: {type(e).__name__}: {e}")
        continue
    if cands is None or len(cands) == 0:
        continue
    for _, row in cands.iterrows():                       # round levers to whole units; keep flips that survive
        cand, changed = q.copy(), {}
        for c in actionable:
            new = min(max(int(round(float(row[c]))), int(LEVERS[c][0])), int(LEVERS[c][1]))
            old = int(round(float(q.iloc[0][c])))
            if new != old:
                cand.iat[0, cand.columns.get_loc(c)] = new
                changed[c] = (old, new)
        if 1 <= len(changed) <= 3 and best.predict_proba(cand[gnames].astype(float))[0, 1] >= 0.5:
            found = (j, float(p_s[j]), float(best.predict_proba(cand[gnames].astype(float))[0, 1]), changed)
            break
    if found:
        break

if found:
    j, p0, p1, changed = found
    print(f"\nCompany #{j}: predicted fail (p_success={p0:.2f}). Smallest realistic actionable change that flips it:")
    for c, (o, n) in changed.items():
        print(f"  {shorten(back[c])}: {o}  ->  {n}")
    print(f"  => p_success {p0:.2f} -> {p1:.2f}  (crosses the 0.5 decision line)")
else:
    print("No small, realistic, founder-controllable change flips these companies - the model leans on")
    print("features the founder cannot act on. An honest (if unsatisfying) recourse result worth stating.")
```

## 19. Synthesis - same data, same features, four methods

The earlier version of this table compared rankings taken from **different feature sets** (the
logistic / RF / PFI columns still had the recession leak in; the SHAP column could not). Part of the
"disagreement" was therefore just the leak appearing in three columns and being structurally absent
from the fourth - a confound, not a finding.

So we recompute all four on the **de-leaked** matrix. Now the only thing that differs is the
*method*: a linear effect, impurity reduction, permutation loss, and Shapley attribution.

```python
# all four importances on the SAME (de-leaked) feature set
r_s = permutation_importance(rf_s, Xte_s, yte, n_repeats=20, random_state=SEED, scoring="roc_auc")
scores = pd.DataFrame({
    "logistic |coef|": pd.Series(np.abs(lr_safe[-1].coef_[0]), index=safe_cols),
    "RF impurity":     pd.Series(rf_s.feature_importances_, index=safe_cols),
    "PFI (test)":      pd.Series(r_s.importances_mean, index=safe_cols),
    "SHAP (LGBM)":     pd.Series(np.abs(sv.values).mean(0), index=safe_cols),
})
rankings = pd.DataFrame({c: scores[c].sort_values(ascending=False).head(6).index.map(shorten)
                         for c in scores.columns})
rankings.index = [f"#{k + 1}" for k in range(len(rankings))]
display(rankings)

rho = scores.corr(method="spearman")
print("\nSpearman rank correlation between the four importance rankings:")
print(rho.round(2).to_string())
overlap = len(set.intersection(*[set(scores[c].nlargest(6).index) for c in scores.columns]))
print(f"\nfeatures in ALL four top-6 lists: {overlap}/6")
```

## 20. Conclusions

- **Interpretability is an audit tool.** The naive forest reached **0.955** test AUC, and PFI put
  `Internet Activity Score`, `google page rank` and the recession flag at the top - hindsight
  features nobody had on day one. The *ranking*, not the score, is what exposed them.

- **But a single held-out number cannot audit leakage.** Removing the post-hoc blocks cumulatively
  moved test AUC 0.955 -> 0.949 -> 0.938 -> **0.943** - the last removal made it go *up*, and the
  bootstrap band on 142 test rows is wider than every one of those gaps. So "leakage is
  distributed" and "the rest is clean" are *both* unfalsifiable with this test set. That is the
  honest result, and it is why the chapter reaches for rankings plus domain judgement instead of
  watching one score move.

- **Preprocess inside the split.** Imputation happens *after* the split, on training medians only
  (**8.9%** of cells were missing). Median-filling the full frame first would have let the test set
  vote on the training data - the same family of mistake as the leaky column, just quieter and
  much easier to ship.

- **A tuned model is not automatically a better one.** Optuna pushed **5-fold CV** AUC to
  **0.970**; the tuned LightGBM then scored **0.927** on test - *below* the de-leaked random
  forest's **0.949**. On 472 rows, an aggressive search buys optimism at least as readily as
  accuracy.

- **Ranking is not calibration.** The tuned model predicts a mean P(success) of **0.704** against
  an actual **0.648** (Brier **0.130**). AUC is invariant to monotone rescaling, so **0.927** is
  blind to that gap. The SHAP waterfall's "probability" and the counterfactual's 0.5 line are
  therefore statements about the model's *score*, not about a startup's real chances.

- **Importance is method-dependent - now with a number.** On one de-leaked feature set, Spearman
  correlation between the four rankings runs from **0.05** (logistic vs PFI) to **0.78** (RF
  impurity vs SHAP), and **0 of 6** features appear in all four top-6 lists. Always say *which*
  importance, on *what* data, for *which* model.

- **Correlation is structural here, not just statistical.** `year of founding` and
  `Age of company in years` are the same fact written twice (**r = 1.00**). Permuting them together
  costs more than permuting them one at a time - the credit-splitting the chapter warns about -
  though at this sample size the whole effect is a fraction of a point of AUC.

- **Important != causal.** Prior-startup experience and social-media activity correlate with
  success; a founder cannot *cause* success by editing those fields. PFI and SHAP report what the
  *model uses*, not what moves the world - which is also why the recourse section restricts DiCE to
  levers a founder actually controls.

- **Caveats.** 472 rows x 134 features is a small-n/high-p regime: PFI error bars are wide, the
  142-row test set cannot resolve 0.01 AUC, and the dataset itself is subjective and messy. Treat
  every "driver" above as a hypothesis to test, not a finding - which is exactly the posture
  interpretability is meant to enforce.
